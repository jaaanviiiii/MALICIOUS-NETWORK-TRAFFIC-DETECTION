```
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Python data science libraries imported successfully!")
```

### 1. Raw Dataset Inspection
The CIC-IDS2017 dataset consists of 8 separate CSV captures collected over 5 days of normal and malicious traffic.


```
raw_dir = os.path.join("..", "dataset", "raw")
csv_files = sorted(glob.glob(os.path.join(raw_dir, "*.csv")))
print(f"Total raw CSV files found: {len(csv_files)}\n")

for f in csv_files:
    size_mb = os.path.getsize(f) / (1024 * 1024)
    print(f"- {os.path.basename(f):55s} | Size: {size_mb:6.1f} MB")
```

### 2. Loading the Cleaned Preprocessed Sample
The raw files total over 2.8 million records. A stratified sample of ~129,000 flows was created to preserve all rare attack types while enabling fast, responsive training.


```
processed_path = os.path.join("..", "dataset", "processed", "cicids2017_cleaned_sample.csv")
df = pd.read_csv(processed_path)
print(f"Dataset Dimensions: {df.shape[0]:,} rows x {df.shape[1]} columns")
df.head()
```

### 3. Target Distribution Analysis
We evaluate both binary classification (Safe vs Malicious) and multiclass classification (Attack Categories).


```
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Binary Target
sns.countplot(data=df, x="Is_Malicious", ax=axes[0], palette="Blues_r")
axes[0].set_title("Binary Target Distribution (0: Benign, 1: Malicious)", fontsize=12, fontweight="bold")
axes[0].set_xticklabels(["0: Benign", "1: Malicious"])

# Multiclass Target
order = df["Attack_Category"].value_counts().index
sns.countplot(data=df, y="Attack_Category", order=order, ax=axes[1], palette="viridis")
axes[1].set_title("Attack Category Breakdown", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.show()
```

### 4. Key Preprocessing Decisions (Unit 1 Summary)
1. **Column Formatting:** Stripped all leading/trailing whitespace from column names.
2. **Infinite Values:** Replaced `inf` and `-inf` with feature medians.
3. **Duplicate Features:** Removed `'Fwd Header Length.1'`.
4. **Zero-Variance Features:** Dropped constant columns that provided zero information gain.
5. **Stratification:** Maintained rare classes (Heartbleed, SQL Injection, Infiltration).
