### 7. Academic Observations & Conclusions
1. **Random Forest** achieved the highest overall accuracy (99.77%) and ROC-AUC (99.99%) by averaging 100 decorrelated trees.
2. **Decision Tree** performed remarkably well (99.33%) but exhibits high variance if unconstrained (memorizing at depth > 20).
3. **Support Vector Machine (LinearSVC)** achieved 92.55% accuracy and 97.40% ROC-AUC, providing a solid maximum-margin linear boundary.
4. **Logistic Regression** achieved 93.39% accuracy, while **Gaussian Naive Bayes** scored 64.46% due to high correlation among flow features.
5. **Minority Class Challenge:** Rare attacks like Web Attacks and Infiltration show lower recall in multiclass mode, demonstrating that payload-level attacks need deep packet inspection.
